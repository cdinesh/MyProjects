```python
import pandas as pd
import numpy as np
```

```python
airline = pd.read_csv('../data/Airline.csv')
airports = pd.read_csv('../data/US_Airports.csv') # With Latitude and Longitude
fares = pd.read_csv('../data/AverageFares.csv')  # Average fares to fly at each airport
zillow = pd.read_csv('../data/Zilliow_Value.csv')
lax_df = pd.read_csv('../data/LAX_Listings.csv')
```

```python
#lax['neighbourhood'].unique()
```

```python
columns_to_drop = ['name', 'license', 'host_name', 'id', 'host_id', 'calculated_host_listings_count']

##Thresholds
#Number of nights we determine if it's vacation vs short term
long_stay_threshold = 14 
#Filtering out low traffic areas with low numbers of review
no_review_threshold = 10


#Drop irrelevant columns and NA
lax = lax_df.drop(columns = columns_to_drop).dropna()

```

Remove neighbourhoods with less than 100 listings

```python
#Get neighborhoods with 100+ listings
popular_neighbourhoods = (
    lax['neighbourhood'].value_counts()[lambda x: x > 100].index
)

#Filter the original DataFrame
lax = lax[lax['neighbourhood'].isin(popular_neighbourhoods)]
print(lax.head())
```

Remove data with no recent reviews, reviews below the current threshold (10), and create a binary column on if it's long stay or short stay (14 days)

```python
lax = lax[lax['last_review'] > '2024-01-01']
lax = lax[lax['number_of_reviews_ltm'] > no_review_threshold]
lax = lax.reset_index(drop=True)

#If long stay (min nights > 14 days) set to 1, if short stay, set to 0
lax['long_stay'] = np.where(lax['minimum_nights'] > long_stay_threshold, 1, 0)

lax
```

```python
lax = lax.groupby(['neighbourhood', 'latitude','longitude', 'long_stay', 'room_type']).agg({
    'price': 'mean',
    'availability_365': 'mean',
    'room_type': 'count'  # counts the number of listings per group
}).rename(columns={'room_type': 'listing_count'}).round(1)

lax = lax.reset_index()
#Adding Los Angeles so that we can join it with the airport data1
lax['airport_city'] = 'Los Angeles'
lax['airport_code'] = 'LAX'

lax

```

```python
# Process Airline data 
Restrict to LAX airport
airport_df = airline[airline['ORIGIN'] == 'LAX']
fares_df = fares[fares['Airport Code'] == 'LAX']
# Aggregating passenger count for each airport
passenger_count = airport_df.groupby('ORIGIN')['PASSENGERS'].sum().reset_index()

# Merge passenger count with average fares
airport_data = pd.merge(passenger_count, fares_df, left_on='ORIGIN', right_on='Airport Code')
airport_data = airport_data[['Airport Code','Airport Name','City Name','State Name', 'PASSENGERS', 'Average Fare ($)']] 
airport_data.columns = ['airport_code','airport_name', 'airport_city', 'airport_state','passenger_count','average_fare']
print(airport_data)
```

```python
import geopy.distance
import matplotlib.pyplot as plt
import seaborn as sns

lax_location = (33.9416, -118.4085)  # LAX latitude and longitude. We can use US_Airports.csv for other airport locations
# Use Haversine Formula to calculate distance between lat and lon
# Ref https://en.wikipedia.org/wiki/Haversine_formula
# geodesic distnace 
# https://stackoverflow.com/questions/19412462/getting-distance-between-two-points-based-on-latitude-longitude#:~:text=import%20mpu%20%23%20Point%20one%20lat1,(dist)%20%23%20gives%20278.45817507541943.


def calculate_distance(row):
    listing_location = (row['latitude'], row['longitude'])
    return geopy.distance.geodesic(lax_location, listing_location).miles

lax['distance_from_airport'] = lax.apply(calculate_distance, axis=1)

# Merge the aggregated passenger data with the LAX Listings 
lax_listings = lax.merge(airport_data, how='left', left_on='airport_code', right_on='airport_code')
#print(lax_listings.head())
```
